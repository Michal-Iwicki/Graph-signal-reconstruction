# Optional External Dataset Downloads

These cells only download the external datasets required by the real-data experiments. They do not preprocess the data, build the graph, or run the reconstruction scripts. `kagglehub` is listed in `requirements.txt`, and Kaggle authentication may be required. After downloading, verify the expected folder names and file names before running the experiments: MovieLens expects `data/movie/ml-latest-small/ratings.csv` (or a path passed via `--ratings-path`), while METR-LA expects both `data/metr/METR-LA.h5` and `data/metr/adj_METR-LA.pkl`.

```python
import kagglehub

# Download the MovieLens 100K dataset.
path = kagglehub.dataset_download("abhikjha/movielens-100k",output_dir="./data/movie",force_download=True)
print("Path to dataset files:", path)
```

## METR-LA Dataset

Run this cell to download the external traffic dataset into `data/metr/`. The METR-LA loader reads `METR-LA.h5` as the timestamp-by-sensor traffic matrix and `adj_METR-LA.pkl` as the adjacency data; both exact filenames must be present in that directory. This cell only downloads and extracts the dataset; it does not start training or reconstruction.

```python
import kagglehub

# Download the METR-LA traffic dataset.
path = kagglehub.dataset_download("annnnguyen/metr-la-dataset",output_dir="./data/metr",force_download=True)

print("Path to dataset files:", path)
```
